```python
%load_ext autoreload
%autoreload 2

import numpy as np
import pandas as pd

import frb_isotropy as iso
import frb_robustness as rob
```

```python

# ==========================================================
# Explicit analysis configuration
# ==========================================================

RUN_CONFIG = iso.AnalysisConfig(
    name="fiducial",

    use_gal_mask=True,
    use_sel_func=True,
    n_jobs=100,

    gal_cut=20.0,

    min_sep=0.1,
    max_sep=180.0,
    bin_size=10.0,

    coarse_bins=tuple(np.arange(0, 181, 20)),

    nside_sf=64,
    smooth_sigma=3.0,
    perturbation_scale=1.0,

    svd_eigenvalue_cut=1e-2,

    n_rand_factor=20,
    n_ensemble=20,
    n_mocks_per_ensemble=50,

    nside_jackknife=4,
    min_jackknife_regions=20,

    n_bootstrap=500,

    overlap_radius_deg=5.0,
    overlap_nside=32,
)

iso.apply_analysis_config(RUN_CONFIG)
```

```python
# ==========================================================
# Mini robustness grid
# ==========================================================

mini = rob.run_full_robustness_scan(

    gal_cut_range=[15.0, 20.0],

    bin_size_range=[5.0, 10.0],

    nside_sf_range=[64],

    smooth_sigma_range=[3.0, 7.5],

    n_jobs=100,

    output_csv=(
        "outputs/robustness_mini.csv"
    ),
)


mini[
    [
        "gal_cut",
        "bin_size",
        "nside_sf",
        "smooth_sigma",
        "p_chi2_empirical",
        "p_chi2_svd_empirical",
        "sigma_svd_empirical",
        "empirical_pvalue",
        "covariance_condition",
        "effective_modes",
    ]
]

mini.describe().T
```
